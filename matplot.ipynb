```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.linspace(0,5,11)
x
y=x**2
y
```

```python
plt.plot(x,y)
plt.title('Square Numbers')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
```

```python
plt.subplot(2,2,1)
```

object orienteed way

```python
fig=plt.figure(figsize=(8,4), dpi=500)
axis1=fig.add_axes([0.1,0.1,0.8,0.8])
axis1.plot(x,y)
axis2=fig.add_axes([0.2,0.5,0.4,0.3])
axis2.plot(y,x) 
plt.savefig('myfig.png', dpi=200)
```
