```python
import pandas as pd
import numpy as np

data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

```python
df.dropna(thresh=3)
```

```python
df.fillna(0)
```

```python
value={'A': 0, 'B': 1, 'C': 2, 'D': 3}
df.fillna(value=value)
```

```python
import pandas as pd

# DataFrame 1: Employee information
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DataFrame 2: Salary information
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees, salaries, on='employee_id', how='inner')
```

```python
pd.merge(employees, salaries, on='employee_id', how='outer')
```

```python
pd.merge(employees, salaries, on='employee_id', how='left')
```

```python
pd.merge(employees, salaries, on='employee_id', how='right')
```

```python
import pandas as pd

df = pd.DataFrame({
    "Department": ["HR", "IT", "IT", "Finance", "HR", "Finance"],
    "Employee": ["John", "Anna", "Peter", "Bob", "Linda", "Tom"],
    "Salary": [50000, 70000, 65000, 80000, 55000, 90000],
    "Experience": [2, 5, 4, 8, 3, 10]
})

print(df)
```

```python
df.groupby("Department")["Salary"].sum()
```

```python
df.groupby("Department")["Salary"].mean()

```

```python
df.groupby("Department")["Salary"].agg(["sum", "mean", "min", "max", "count"])
```
