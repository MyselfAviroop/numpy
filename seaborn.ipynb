HEATMAP

```python
tipscorr=tips[['total_bill','tip','size']]
```

```python
tipscorr.corr()
```

```python
sns.heatmap(tipscorr.corr(), annot=True, linewidths=0.5)
```

CLUSTERMAP

```python
sns.clustermap(tipscorr.corr(), annot=True, linewidths=0.5)
```
